```python
import plotly
import plotly.graph_objs as go
import plotly.express as px
from plotly.subplots import make_subplots
from scipy.optimize import fsolve
import numpy as np
import pandas as pd

x = np.arange(-2*np.pi,2*np.pi, 0.1)
def f(x):
    return np.sin(x)
def g(x):
    return 1-np.cos(x)
```

```python

intersections_x=[-2*np.pi, -1.5*np.pi, 0, 0.5*np.pi]
intersections_y = [f(xi) for xi in intersections_x]
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=f(x), name='f(x)=sin(x)'))
fig.add_trace(go.Scatter(x=x, y=g(x),name='g(x) = 1 - cos(x)'))
# Точки пересечения
fig.add_trace(go.Scatter(x=intersections_x, y=intersections_y, mode='markers', name='Пересечения', marker=dict(size=10, color='red')))
fig.update_layout(legend_orientation="h",
                  legend=dict(x=.5, xanchor="center"),
                  title="Графики функций",
                  title_x=0.5,
                  xaxis_title="x",
                  yaxis_title="y",
                  margin=dict(l=0, r=0, t=30, b=0))
fig.show()
```
