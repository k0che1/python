```python
import plotly
import plotly.graph_objs as go
import plotly.express as px
from plotly.subplots import make_subplots
from scipy.optimize import fsolve
import numpy as np
import pandas as pd
import math

x1 = np.arange(-3,0.1, 0.1)
x2 = np.arange(0.01,3.01, 0.1)
x=np.arange(-3,3.01, 0.1)
def log_scale(x):
    return math.log10(x)
def f1(x):
  return -10**-x
def f2(x):
  return 10**x
```

```python
#Задание 1
fig = go.Figure()
fig.add_trace(go.Scatter(x=x1, y=f1(x1), line=dict(color='red'), name='x<0'))
fig.add_trace(go.Scatter(x=[0], y=[0], name='x=0'))
fig.add_trace(go.Scatter(x=x2, y=f2(x2), line=dict(color='red'), name='x>0'))
fig.update_layout(
    title="График функций",
    title_x=0.5,
    #yaxis_range=[-2,2],
    xaxis_title="x",
    yaxis_title="y"
)
fig.show()
```

```python
#Задание 2
fig = go.Figure()
fig.add_trace(go.Scatter(x=x1, y=f1(x1), line=dict(color='red'), name='x<0'))
fig.add_trace(go.Scatter(x=[0], y=[0], name='x=0'))
fig.add_trace(go.Scatter(x=x2, y=f2(x2), line=dict(color='red'), name='x>0'))

fig.update_layout(
    title="График функций",
    title_x=0.5,
    #yaxis_range=[-2,2],
    xaxis_title="x",
    yaxis_title="y"
)
fig.update_yaxes(type="log")
fig.show()
```

```python
#Задание 3

y1 = f1(x1)
log_y1 = np.log10(np.abs(y1))
sign_y1 = np.sign(y1)
y1_display = sign_y1 * log_y1

y2 = f2(x2)
log_y2 = np.log10(np.abs(y2))
sign_y2 = np.sign(y2)
y2_display = sign_y2 * log_y2

fig = go.Figure()
fig.add_trace(go.Scatter(x=x1, y=y1_display, line=dict(color='red'), name='x<0', hovertemplate =
    'x= %{x:.2f}'+'<br>y= %{y:.2f}'))
fig.add_trace(go.Scatter(x=[0], y=[0], name='x=0', hovertemplate =
    'x= 0 <br>y= 0'))
fig.add_trace(go.Scatter(x=x2, y=y2_display, mode='lines', line=dict(color='red'), name='x>0', hovertemplate =
    'x= %{x:.2f}'+'<br>y= %{y:.2f}'))

fig.update_xaxes(showspikes=True, spikecolor="green", spikesnap="cursor", spikemode="across")
fig.update_yaxes(showspikes=True, spikecolor="orange", spikethickness=2)
fig.update_layout(
    yaxis=dict(
        title='sign(f(x)) * log10(|f(x)|)',
        tickvals=[-4, -3, -2, -1, 0, 1, 2, 3, 4],
        ticktext=['$$-10^4$$', '-1000', '-100', '-10', '0', '10', '100', '1000', '$$10^4$$']
    ),
    title="График функций",
    title_x=0.5,
    xaxis_title="x",
    yaxis_title="y"
)

fig.show()
```
