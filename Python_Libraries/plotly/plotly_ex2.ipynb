```python
import plotly
import plotly.graph_objs as go
import plotly.express as px
from plotly.subplots import make_subplots
from scipy.optimize import fsolve
import numpy as np
import pandas as pd

def f(x):
  return np.sin(x/5) * np.exp(x/10) + 5 * np.exp(-x/2)
def g(x):
  return 1-x**3+x**5
```

```python
x = np.linspace(0,500, 200)

fig = make_subplots(
    rows=6, cols=2,
    column_widths=[3, 1],
    specs=[[{"rowspan": 4}, {}],
           [None,{}],
           [None,{}],
           [None,{}],
           [{"rowspan": 2},{}],
           [{},{}]])

fig.add_trace(go.Scatter(x=x, y=f(x), line=dict(color='red')), row=1, col=1)
fig.add_trace(go.Scatter(x=x, y=g(x), line=dict(color='green')), row=1, col=1)

fig.add_trace(go.Scatter(x=x, y=f(x), line=dict(color='red')), row=1, col=2)
fig.add_trace(go.Scatter(x=x, y=g(x), line=dict(color='green')), row=1, col=2)
fig.update_xaxes(range=[430,450], row=1, col=2)

fig.add_trace(go.Scatter(x=x, y=f(x), line=dict(color='red')), row=2, col=2)
fig.add_trace(go.Scatter(x=x, y=g(x), line=dict(color='green')), row=2, col=2)
fig.update_xaxes(range=[450,465], row=2, col=2)

fig.add_trace(go.Scatter(x=x, y=f(x), line=dict(color='red')), row=3, col=2)
fig.add_trace(go.Scatter(x=x, y=g(x), line=dict(color='green')), row=3, col=2)
fig.update_xaxes(range=[465,475], row=3, col=2)

fig.add_trace(go.Scatter(x=x, y=f(x), line=dict(color='red')), row=4, col=2)
fig.add_trace(go.Scatter(x=x, y=g(x), line=dict(color='green')), row=4, col=2)
fig.update_xaxes(range=[475,490], row=4, col=2)

fig.add_trace(go.Scatter(x=x, y=f(x), line=dict(color='red')), row=5, col=1)
fig.add_trace(go.Scatter(x=x, y=g(x), line=dict(color='green')), row=5, col=1)
fig.update_yaxes(type="log", row=5, col=1)

fig.update_traces(showlegend=False)
fig.update_layout(height=900, width=1200)
fig.show()
```
