```python
import plotly
import plotly.graph_objs as go
import plotly.express as px
from plotly.subplots import make_subplots
from scipy.optimize import fsolve
import numpy as np
import pandas as pd
import math

df=pd.read_csv('Общ.csv', sep=';',decimal=',')
```

```python
print(len(df))
col=['green' if df.iloc[0, i]>0 else 'red' for i in range(df.shape[1])]
df.loc[len(df)] = col
print(df)
```

```python
fig = go.Figure()
fig.update_xaxes(range=[-1,29])
fig.update_yaxes(zeroline=True, zerolinewidth=1, zerolinecolor='black')

fig.add_trace(go.Scatter(x=df.columns, y=df.iloc[0],

                         mode='lines+markers',
                         hovertemplate =
                        '%{x}<br>Прирост населения:%{y}',
                         marker=dict(color=df.iloc[1],size=10)
                         ))


fig.update_layout(hoverlabel_align = 'left',
                  title="Прирост населения по годам",
                  xaxis_title="Год",
                  yaxis_title="Прирост населения",
                  showlegend=False)

fig.show()
```
